```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 150, 130, 180, 200]

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
import numpy as np
import statistics

marks = [12, 15, 15, 18, 20, 22, 25]

print("Mean =", np.mean(marks))
print("Median =", np.median(marks))
print("Mode =", statistics.mode(marks))

```

```python
import numpy as np

data = [2, 4, 4, 4, 5, 5, 7, 9]

print("Mean =", np.mean(data))
print("Population Variance =", np.var(data))
print("Standard Deviation =", np.std(data))
```

```python
import numpy as np

study_hours = [1, 2, 3, 4, 5, 6]
marks = [45, 50, 55, 65, 70, 80]

covariance = np.cov(study_hours, marks, ddof=0)[0, 1]
correlation = np.corrcoef(study_hours, marks)[0, 1]

print("Covariance =", covariance)
print("Correlation =", correlation)
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

```

```python
from scipy import stats

sample = [28, 30, 31, 29, 32, 30, 27, 31, 29, 30]

# H0: Mean = 30
t_stat, p_value = stats.ttest_1samp(sample, 30)

print("t-statistic =", t_stat)
print("p-value =", p_value)

alpha = 0.05

if p_value <= alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```
